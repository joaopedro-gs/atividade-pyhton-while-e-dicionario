```python
# como funciona o while
# while condição:

# exemplo 1 - contar até o usuario pedir

limite = int(input("informe um numero limite"))

inicio = 0

while inicio <= limite:
  print(inicio)
  inicio = inicio + 1
```

```python
# controle com while
senha_correta = "abc123"
tentativas = 0

while tentativas < 3:
  senha = input("Digite sua senha")
  if senha_correta == senha:
    print("Bem vindo ao sistema")
    break #quebrar
  else:
    print("senha incorreta")
    tentativas = tentativas + 1
    print(f"numero de tentativas = {tentativas}")
    if tentativas == 3:
      print("senha bloqueada")
```

```python
# while true

while True:
  comando = input("para sair, digite SAIR")
  if comando.lower() == "sair":
    print("até mais")
    break
  else:
    print("continuando o sistema")
```

```python
# ATIVIDADE 1 - FACIL
# Quero um sistema que pergunte ao usuario se quer salvar uma informaçao em uma lista de carros
# a lista ---> carros = ["uno", "corola", "fusca"]
# se o usuario quer colocar um novo carro, vai adicionar, se ele digitar sair, finaliza o sistema

carros = ["uno", "corola", "fusca"]

while True:

  novo_carro = input("Digite o nome de um carro ou 'sair' para finalizar")

  if novo_carro.lower() == "sair":
    print("Sistema finalizado!")
    break

  else:
    carros.append(novo_carro)
    for i in carros:
      print(i)

  print("Carro adicionado com sucesso!")

print("Lista de carros: ", carros)

# ATIVIDADE 2 - FACIL PRA MÉDIO
```
